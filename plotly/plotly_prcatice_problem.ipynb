```python
# 1. Write a program to draw a scatter plot for random 1000 x and y coordinates 
import numpy as np
import pandas as pd
import plotly.express as ple

data={
    'x':np.random.randint(1,1000,1000),
    'y':np.random.randint(1,1000,1000)
}

df=pd.DataFrame(data)

fig=ple.scatter(df,x='x',y='y',title="Scatter plot for random 1000 x and y coordinates")
fig.show()

```

```python
# 2. Write a program to draw line and scatter plots for random 100 x and y coordinates 

data={
    "X":np.random.randint(1,100,100),
    "Y":np.random.randint(1,100,100)
}

df=pd.DataFrame(data)

line_fig=ple.line(df,x='X',y='Y')

scatter_fig=ple.scatter(df,x='X',y='Y')

line_fig.add_traces(scatter_fig.data)

line_fig.update_layout(xaxis_title='X axis',yaxis_title='Y axis',title="line and scatter plots for random 100 x and y")

line_fig.show()
```

```python
# 3. Write a program to draw a scatter plot for random 500 x and y coordinates and style it 

df=pd.DataFrame({
    "X":np.random.randint(1,1000,500),
    "Y":np.random.randint(1,1000,500)
})

fig=ple.scatter(df,x="X",y="Y",color='Y',size='X',size_max=10,opacity=0.7,title="scatter plot for random 500 x and y")

fig.update_traces(marker=dict(line=dict(width=2)))
fig.update_layout(
    xaxis_title='X axis',
    yaxis_title='Y axis'
)

fig.show()
```

```python
# 4. Write a program to draw a scatter plot for a given dataset and show datalabels on hover 
df=pd.read_csv("https://raw.githubusercontent.com/plotly/datasets/master/2014_usa_states.csv")

fig=ple.scatter(df,x="Postal",y="Population",title="US State Population",hover_name="State",hover_data={"Postal":True,"Population":True})

fig.update_layout(
    xaxis_title="State Code",
    yaxis_title="Population"
)
fig.show()
```
